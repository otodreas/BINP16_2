```python
class DNASequence:
    def __init__(self, name, seq):
        # self: the INSTANCE of the class
        self.name = name
        self.seq = seq
    # function to compute another attribute (aka METHOD)
    def gc(self):
        g = self.seq.count('G')
        c = self.seq.count('C')
        return (g+c) / len(self.seq)
        
dna = DNASequence('gene1', 'ATGCGT')
print(dna.gc())
```

```python
class DNASequence:
    # Assign attributes
    def __init__(self, name, seq):
        self.name = name
        self.seq = seq

    # Format print (make readable)
    def __str__(self):
        return f'DNA Sequence. Name: {self.name}, Sequence: {self.seq}'

    def __len__(self):
        return len(self.seq)

    def clean_dna(self):
        """Returns an uppercase DNA string with only A, C, G, T, N. Other characters, including spaces, become 'N'."""
        allowed = {'A', 'C', 'G', 'T', 'N'}  # Valid DNA bases, including N
        cleaned = []  # Build result in a list
        for char in self.seq.upper():  # Convert to uppercase
            cleaned.append(char if char in allowed else 'N')  # Keep valid, else N
        return ''.join(cleaned)  # Join to string    

        
    def is_valid(self):
        valid_chars = set('ACGT')
        chars = set(self.seq.upper())
        if valid_chars | chars == valid_chars: # only 'AGCT' must be contained across the sets
            return True
        else:
            return False
            
    def reverse(self):
        if self.is_valid():
            return self.seq[::-1]
        
        
    def complement(self):
        complement_seq = []
        pairs = [['a', 't'], ['c', 'g']]
        for b in self.seq:
            if b in pairs[0]:
                comp = pairs[0][not pairs[0].index(b)]
            else:
                comp = pairs[1][not pairs[1].index(b)]
            complement_seq.append(comp)
        return ''.join(complement_seq)
        
    # You can call a function upstream in the class inside another method
    def reverse_complement(self):
        return self.complement()[::-1]


    def base_freq(self):
        bcounts = {'A': 0, 'C': 0, 'G': 0, 'T': 0}
        for b in self.seq.upper():
            bcounts[b] += 1
        return bcounts

    # do a motif search for motif 'subseq'. when using this method, only pass stuff right of self!
    def has_subseq(self, subseq):
        if subseq in self.seq:
            return True
        else:
            return False

    def to_upper(self):
        return self.seq.upper()

    def to_rna(self): # this could be smoother if i standardized case earlier
        rna = self.seq.replace('t', 'u')
        rna = rna.replace('T', 'U')
        return rna

    def count_kmers(self, k):
        """Returns dictionary of k-mer frequencies (overlapping, skips N)."""
        if k <= 0:  # Validate k
            raise ValueError("k must be positive")
        counts = {}  # Store k-mer counts
        for i in range(len(self.seq) - k + 1):  # Slide window
            kmer = self.seq[i:i+k]
            if 'N' in kmer:  # Skip invalid
                continue
            counts[kmer] = counts.get(kmer, 0) + 1  # Increment count
        return counts
        
dna = DNASequence('geneX', 'attac')
print(dna)

print(dna.seq)
print(dna.reverse())
print(dna.complement())
print(dna.reverse_complement())
print(dna.is_valid())
print(len(dna)) # by default, the object type DNASequence does not have len(). we assign it to the length of the sequence above. Try commenting out the def __len__() and see
# use methods that return stuff like dictionaries to define new variables
counts = dna.base_freq()
print(counts, type(counts))
print(dna.has_subseq('tta'))
print(dna.to_upper())
print(dna.to_rna())
print(dna.count_kmers(3))
```
